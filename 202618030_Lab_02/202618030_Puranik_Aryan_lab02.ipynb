<h3>Vectorized Arithmetic and Linear Algebra</h3>

```python

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)




addition = A + B
print("Matrix Addition")
print("\nA + B:")
print(addition)



elementwise_multiplication = A * B

print("\nElement-wise multiplication (A * B):")
print(elementwise_multiplication)



matrix_multiplication = A @ B
print("Matrix Multiplication")
print("\nMatrix multiplication (A @ B):")
print(matrix_multiplication)



transpose_A = A.T

print("\nTranspose of A:")
print(transpose_A)


# ---------------------------------------------------------
# 6. Determinant
# ---------------------------------------------------------

det_A = np.linalg.det(A)

print("\nDeterminant of A:")
print(det_A)



#  Check whether A is invertible

if not np.isclose(det_A, 0):
    
    # Find inverse
    A_inverse = np.linalg.inv(A)

    print("\nInverse of A:")
    print(A_inverse)

    # Verify A @ A^-1 = I
    identity = A @ A_inverse

    print("\nA @ A^-1:")
    print(identity)

    # Verify using np.allclose()
    verification = np.allclose(identity, np.eye(A.shape[0]))

    print("\nIs A @ A^-1 approximately equal to Identity?")
    print(verification)

else:
    print("\nA is not invertible because its determinant is zero.")
```

<h3>Filtering and Querying</h3>

```python
import pandas as pd

# If you haven't loaded the dataset:
# df = pd.read_csv("Titanic-Dataset.csv")

# ---------------------------------------------------------
# 1. Male passengers older than 50
# ---------------------------------------------------------

male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("1. Male passengers older than 50:", len(male_over_50))


# ---------------------------------------------------------
# 2. Female first-class passengers
#    How many, and what percentage survived?
# ---------------------------------------------------------

female_first = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

female_first_count = len(female_first)

female_first_survival_percentage = (
    female_first["Survived"].mean() * 100
)

print("\n2. Female first-class passengers:", female_first_count)
print("   Survival percentage:",
      female_first_survival_percentage, "%")


# ---------------------------------------------------------
# 3. Age 20-40, Fare above overall median, and survived
# ---------------------------------------------------------

overall_fare_median = df["Fare"].median()

age_fare_survived = df[
    (df["Age"].between(20, 40)) &
    (df["Fare"] > overall_fare_median) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, fare above median, survived:",
      len(age_fare_survived))


# ---------------------------------------------------------
# 4. Travelling alone, age < 30, and did not survive
# ---------------------------------------------------------

alone_under_30_not_survived = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Alone, age < 30, did not survive:",
      len(alone_under_30_not_survived))


# ---------------------------------------------------------
# 5. Embarked = S, Pclass 2 or 3,
#    Fare above Southampton median
# ---------------------------------------------------------

southampton_median = df.loc[
    df["Embarked"] == "S", "Fare"
].median()

southampton_condition = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("\n5. Embarked S, Pclass 2/3, fare above Southampton median:",
      len(southampton_condition))
```

<h3>Groupby and Aggregation</h3>

```python
# 1. Survival rate by Sex
survival_by_sex = df.groupby('Sex')['Survived'].mean()
print("--- Survival Rate by Sex ---")
print(survival_by_sex)


# 2. Survival rate by Pclass
survival_by_pclass = df.groupby('Pclass')['Survived'].mean()
print("\n--- Survival Rate by Pclass ---")
print(survival_by_pclass)


# 3. Average Age and Fare by Pclass
# Passing a list of columns ['Age', 'Fare'] to aggregate both
age_fare_by_pclass = df.groupby('Pclass')[['Age', 'Fare']].mean()
print("\n--- Average Age and Fare by Pclass ---")
print(age_fare_by_pclass)


# 4. Passenger count and survival rate by Sex-Pclass
# Passing a list ['Sex', 'Pclass'] to groupby creates a MultiIndex
stats_by_sex_pclass = df.groupby(['Sex', 'Pclass']).agg(
    Passenger_Count=('Survived', 'count'),
    Survival_Rate=('Survived', 'mean')
)
print("\n--- Count and Survival Rate by Sex & Pclass ---")
print(stats_by_sex_pclass)


# 5. Passenger count, average Fare, and survival rate by Embarked
stats_by_embarked = df.groupby('Embarked').agg(
    Passenger_Count=('PassengerId', 'count'), # Can count any non-null column
    Avg_Fare=('Fare', 'mean'),
    Survival_Rate=('Survived', 'mean')
)
print("\n--- Stats by Embarked ---")
print(stats_by_embarked)
```

<h3>Missing Values and Outliers</h3>

```python
# 1. Missing Value Count, Percentage, & Plot

# Calculate missing counts and percentages
missing_counts = df.isnull().sum()
missing_percentage = (missing_counts / len(df)) * 100

# Combine into a single DataFrame for easy viewing
missing_stats = pd.DataFrame({
    'Missing_Count': missing_counts,
    'Percentage(%)': missing_percentage
})

print("--- Missing Values Summary ---")
print(missing_stats[missing_stats['Missing_Count'] > 0]) # Show only columns with missing data

# Plot missing counts as a bar chart
missing_counts[missing_counts > 0].plot(kind='bar', color='salmon', edgecolor='black')
plt.title('Count of Missing Values per Column')
plt.ylabel('Number of Missing Values')
plt.xlabel('Columns')
plt.xticks(rotation=45)
plt.show()


# ==========================================
# 2. Fill Missing Age with Mean
# ==========================================

print("\n--- Imputing Age with Mean ---")
print(f"Missing Age values BEFORE: {df['Age'].isnull().sum()}")

# Fill missing Age values with the mean
mean_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(mean_age)

print(f"Missing Age values AFTER: {df['Age'].isnull().sum()}")


# ==========================================
# 3. Demonstration of Imputation Methods
# ==========================================
# (Creating a dummy copy to demonstrate without altering the cleaned Age column)
df_demo = pd.read_csv('titanic_train.csv') 

# A. Mean Imputation (Best for normally distributed numerical data)
df_demo['Age_Mean'] = df_demo['Age'].fillna(df_demo['Age'].mean())

# B. Median Imputation (Best for skewed numerical data, less sensitive to outliers)
df_demo['Age_Median'] = df_demo['Age'].fillna(df_demo['Age'].median())

# C. Mode Imputation (Best for categorical data like 'Embarked')
# mode() returns a Series, so we use [0] to grab the first/most frequent value
df_demo['Embarked_Mode'] = df_demo['Embarked'].fillna(df_demo['Embarked'].mode()[0])

# D. Random Value Imputation (Maintains original variance)
# We generate random samples from the existing non-null values to fill the NaNs
non_null_ages = df_demo['Age'].dropna()
random_samples = np.random.choice(non_null_ages, size=df_demo['Age'].isnull().sum())

# Create a copy of the column and assign the random values to the missing indices
df_demo['Age_Random'] = df_demo['Age'].copy()
df_demo.loc[df_demo['Age'].isnull(), 'Age_Random'] = random_samples


# ==========================================
# 4. Outlier Detection for 'Fare' (IQR Method)
# ==========================================

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

# Calculate Interquartile Range (IQR)
IQR = Q3 - Q1

# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Count how many fares fall outside these bounds
outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
outlier_count = len(outliers)

print("\n--- Fare Outlier Statistics ---")
print(f"Q1 (25th Percentile): {Q1:.2f}")
print(f"Q3 (75th Percentile): {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
print(f"Total Outliers Count: {outlier_count}")
```

<h3>Features and Pivot table</h3>

```python

# ==========================================
# 1. Feature Engineering: FamilySize & IsAlone
# ==========================================

# Create FamilySize (Siblings/Spouses + Parents/Children + 1 for the passenger themselves)
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Create IsAlone (1 if FamilySize is 1, otherwise 0)
# .astype(int) converts the Boolean True/False into 1/0
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

print("--- New Features (First 5 Rows) ---")
print(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head())


# Pivot Table & Finding Extremes

# Create the pivot table
# aggfunc='mean' is the default for pivot_table, but it's good practice to be explicit
pivot_survival = df.pivot_table(
    values='Survived', 
    index='Sex', 
    columns='Pclass', 
    aggfunc='mean'
)

print("\n--- Survival Rate Pivot Table ---")
print(pivot_survival)


# Unstack the pivot table into a 1D Pandas Series to easily find global max and min
stacked_pivot = pivot_survival.unstack()

# idxmax() and idxmin() return the index labels (Pclass, Sex) of the max/min values
highest_group = stacked_pivot.idxmax()
highest_rate = stacked_pivot.max()

lowest_group = stacked_pivot.idxmin()
lowest_rate = stacked_pivot.min()

print(f"\n--- Highest and Lowest Groups ---")
# highest_group[1] is the Sex, highest_group[0] is the Pclass
print(f"Highest Survival: {highest_group[1].capitalize()}s in Pclass {highest_group[0]} ({highest_rate * 100:.2f}%)")
print(f"Lowest Survival:  {lowest_group[1].capitalize()}s in Pclass {lowest_group[0]} ({lowest_rate * 100:.2f}%)")
```
